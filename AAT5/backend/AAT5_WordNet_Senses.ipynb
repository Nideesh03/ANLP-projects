```python
import json, nltk
from nltk.corpus import wordnet as wn
from nltk.wsd import lesk
import core   # extended_lesk lives in core.py
print('WordNet', wn.get_version())
```

```python
for word in ['bat', 'crane', 'light']:
    syns = wn.synsets(word)
    print(f'\n{word.upper()}: {len(syns)} senses')
    for s in syns[:6]:
        print(f'  {s.name():<18}{s.definition()[:70]}')
    print('  synonyms:', sorted({l.name() for s in syns for l in s.lemmas()} - {word})[:12])
```

```python
contexts = json.load(open('../data/contexts.json'))
for c in contexts:
    s = lesk(nltk.word_tokenize(c['sentence'].lower()), c['word'], pos=c['pos'])
    e = core.extended_lesk(c['word'], c['sentence'], c['pos'])
    print(c['sentence'])
    print('   NLTK lesk    :', s.name(), '-', s.definition()[:60])
    print('   extended lesk:', e['synset'], '-', e['definition'][:60], '| overlap', e['overlap'], '\n')
```
