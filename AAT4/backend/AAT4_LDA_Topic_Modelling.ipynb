```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
docs = open('../data/documents.txt').read().strip().split('\n')
vec = CountVectorizer(stop_words='english')
dtm = vec.fit_transform(docs)
vocab = vec.get_feature_names_out()
print('Documents:', len(docs), '| vocabulary:', len(vocab))
```

```python
for seed in range(10):
    m = LatentDirichletAllocation(n_components=3, random_state=seed, max_iter=50, doc_topic_prior=0.1, topic_word_prior=0.05).fit(dtm)
    print('seed', seed, 'perplexity', round(m.perplexity(dtm), 2))
```

```python
lda = LatentDirichletAllocation(n_components=3, random_state=9, max_iter=50, doc_topic_prior=0.1, topic_word_prior=0.05)
doc_topic = lda.fit_transform(dtm)
for k, w in enumerate(lda.components_, 1):
    p = w / w.sum(); top = p.argsort()[::-1][:5]
    print(f'Topic {k}:', ', '.join(f'{vocab[i]} ({p[i]:.3f})' for i in top))
```

```python
for i, (d, row) in enumerate(zip(docs, doc_topic), 1):
    print(f'D{i:<2} T{row.argmax()+1} {row.max():.2f}  {d[:70]}')
```
