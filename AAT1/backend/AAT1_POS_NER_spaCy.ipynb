```python
import spacy
nlp = spacy.load('en_core_web_sm')
print('spaCy', spacy.__version__, '| pipeline:', nlp.pipe_names)
```

```python
text = open('../data/sample_paragraph.txt').read().strip()
doc = nlp(text)
print(text)
print('\nSentences:', len(list(doc.sents)), '| tokens:', len(doc))
```

Part-of-speech tags

```python
print(f"{'TOKEN':<14}{'UPOS':<7}{'PENN':<6}{'LEMMA':<14}MEANING")
for t in doc:
    if not (t.is_punct or t.is_space):
        print(f'{t.text:<14}{t.pos_:<7}{t.tag_:<6}{t.lemma_:<14}{spacy.explain(t.pos_)}')
```

Named entities

```python
for e in doc.ents:
    print(f'{e.text:<32}{e.label_:<9}{spacy.explain(e.label_)}')
```

```python
from collections import Counter
print('POS frequency :', Counter(t.pos_ for t in doc if not t.is_punct).most_common())
print('Entity labels :', Counter(e.label_ for e in doc.ents))
print('Noun chunks   :', [c.text for c in doc.noun_chunks])
```
